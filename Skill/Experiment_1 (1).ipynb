```python
# 1.Import pandas
import pandas as pd
from google.colab import files

# 2.Upload the CSV file
uploaded = files.upload()
filename = next(iter(uploaded))

# 3.Load the dataset
df = pd.read_csv(filename)

print("Loaded file:", filename)
```

```python
# 4.Display first 5 rows
print("First Five Rows:")
print(df.head())

# 5.Check dataset shape
print("\nDataset Shape:")
print(df.shape)

# 6.Check column names
print("\nColumn Names:")
print(df.columns)

# 7.Check data types
print("\nData Types:")
print(df.dtypes)

# 8.Display dataset information
print("\nDataset Information:")
df.info()

# 9.View statistical summary
print("\nStatistical Summary:")
print(df.describe())

# 10.Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 11.Check duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# 12.Study placement classes
print("\nPlacement Class Distribution:")
print(df["PlacementStatus"].value_counts())

# Calculate class percentages
print("\nPlacement Class Percentage:")
print(df["PlacementStatus"].value_counts(normalize=True) * 100)
```
